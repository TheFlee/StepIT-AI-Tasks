```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Task 1
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [18, 20, 22, 21, 24, 26, 23]
```

```python
line, = plt.plot(days, temperature)
plt.xlabel("Days")
plt.ylabel("Temperature (\u00b0C)")
plt.title("Weekly Temperature")
line.set_marker("o")
line.set_linestyle("--")
plt.grid(True)
for i, t in enumerate(temperature):
    plt.text(i, t + 0.15, str(t), ha="center")
plt.gcf().set_size_inches(10, 5)
plt.show()
```

```python
# Task 2
languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]
```

```python
bars = plt.barh(languages, students, color=np.random.rand(len(languages), 3))

plt.xlabel("Number of Students")
plt.ylabel("Languages")
plt.title("Students by Programming Language")
for i, v in enumerate(students):
    plt.text(v + 0.2, i, str(v), va="center")
max_idx = students.index(max(students))
bars[max_idx].set_edgecolor("black")
bars[max_idx].set_linewidth(3)
plt.gcf().set_size_inches(8, 5)
plt.gca().invert_yaxis()
plt.show()
```

```python
bars2 = plt.bar(languages, students, color=np.random.rand(len(languages), 3))
plt.xlabel("Languages")
plt.ylabel("Number of Students")
plt.title("Students by Programming Language")
for i, v in enumerate(students):
    plt.text(i, v + 0.2, str(v), ha="center")
max_idx = students.index(max(students))
bars2[max_idx].set_edgecolor("black")
bars2[max_idx].set_linewidth(3)
plt.gcf().set_size_inches(8, 5)
plt.show()
```

```python
# Task 3
study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]
ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]
```

```python
fig = plt.figure(figsize=(10, 5))
axes1 = fig.add_axes([0.08, 0.1, 0.8, 0.8])

scatter = axes1.scatter(study_hours, scores, c=ages, cmap="viridis", s=100)
fig.colorbar(scatter, ax=axes1, label="Age")

axes1.set_xlabel("Study Hours")
axes1.set_ylabel("Scores")
axes1.set_title("Study Hours vs Exam Score")
axes1.grid(True)

max_idx = scores.index(max(scores))
axes1.annotate("Highest Score",
               xy=(study_hours[max_idx], scores[max_idx]),
               xytext=(study_hours[max_idx] - 1.5, scores[max_idx] - 3.5),
               arrowprops=dict(arrowstyle="->", color="black"))

axes2 = fig.add_axes([0.12, 0.58, 0.25, 0.25])

axes2.bar(range(1, 11), scores) 
axes2.set_title("Scores", fontsize=8)
axes2.tick_params(labelsize=8)

plt.show()
```
